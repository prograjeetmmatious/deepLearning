ISSUE WITH PERCEPTRON
1. TWO SIMILIAR LINES CLASSIFYING DATA INTO TWO PARTS , CANNOT BE DETERMINED WHICH IS BEST
2. CONVERGENCE
The Perceptron Convergence Theorem states that if the training data is linearly separable, the algorithm will eventually find a decision boundary that correctly classifies all training examples.
1
However, if the data is not linearly separable, the algorithm may continue updating its weights without reaching a stable solution, potentially leading to an endless loop.


MAIN PROBLEM : JUMPING OF LINE IRREGULARLY
That's the key idea behind the Perceptron Convergence Theorem:

$$ \boxed{\text{Linearly separable} \Rightarrow \text{Perceptron converges}} $$

and, roughly,

Not linearly separable⇒Perceptron may never converge

MLP (MULTI LAYER PERCETRON )


*   🧠 Multi-Layer Perceptron (MLP)
Definition

A Multi-Layer Perceptron (MLP) is a neural network made of multiple layers of interconnected neurons, where the output of one layer becomes the input to the next layer.

🧩 Basic Structure
          INPUT LAYER          HIDDEN LAYER          OUTPUT LAYER

             x₁  ──────────────→  ○ h₁  ────────┐
              ╲               ↗       ╲         │
               ╲           ↗            ╲       ↓
             x₂  ──────────────→  ○ h₂  ──────→  ○ ŷ
               ╲           ↘            ╱
                ╲             ↘        ╱
             x₃  ──────────────→  ○ h₃ ────────┘
*   If one neuron produces:

$$ a_1=f(w_1x_1+w_2x_2+b) $$

then another neuron can use \(a_1\) as its input:

$$ z_2=w'_1a_1+w'_2a_2+b' $$

So information flows:

Input→Hidden Layer→Hidden Layer→Output
