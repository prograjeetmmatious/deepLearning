🚀 Forward Propagation in MLP

Forward propagation means:

Taking the input → passing it through each layer → calculating the final output.

The information always moves forward, from input to output.

INPUT              HIDDEN LAYER             OUTPUT

x₁ ───────┐
          ├────→  h₁ ───────┐
x₂ ───────┤                  ├────→ ŷ
          ├────→  h₂ ───────┘
x₃ ───────┘



🔥 Think of each layer as a matrix transformation

Suppose we have 3 input features:

$$ X= \begin{bmatrix} x_1\\ x_2\\ x_3 \end{bmatrix} $$

So:

$$ \boxed{X: 3\times1} $$

Now suppose the hidden layer has 2 neurons.

Its weights are:

$$ W= \begin{bmatrix} w_{11}&w_{12}&w_{13}\\ w_{21}&w_{22}&w_{23} \end{bmatrix} $$

So:

$$ \boxed{W:2\times3} $$

Now multiply:

$$ Z=WX+b $$

Dimensions:

$$ (2\times3)(3\times1) = 2\times1 $$

Therefore:

$$ Z= \begin{bmatrix} z_1\\ z_2 \end{bmatrix} $$
🚨 Look what happened
Input              Hidden Layer

3 values              2 neurons

[x₁]                 [z₁]
[x₂]       ×         [z₂]
[x₃]
3 × 1                 2 × 1

So the representation went:

$$ \boxed{3\text{ values}\rightarrow2\text{ values}} $$

That's probably the "matrix reduction" you're noticing.
